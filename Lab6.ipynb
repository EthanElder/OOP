```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("data.csv")

for x in df.index:
  average_val = (df.loc[x, "Pulse"] + df.loc[x, "Maxpulse"])/2
  df.loc[x, "Average"] = average_val
df.plot()
plt.show()
df.plot(kind='scatter', x='Duration', y='Calories')
plt.show()
```

```python
import pandas as pd
df = pd.read_csv("data01.csv")

med_val = df["median_house_value"].median()
print("Median Value:", med_val)

med_room = df["total_rooms"].median()
for x in df.index:
  if df.loc[x, "total_rooms"] < 500:
    df.loc[x, "total_rooms"] = med_room

low_housing = df[df["housing_median_age"] < 25]
print(low_housing)

df.plot(kind='scatter', x='total_rooms', y='total_bedrooms')
```
